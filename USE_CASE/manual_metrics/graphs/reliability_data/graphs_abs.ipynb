```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Data from your report
data = {
    'Architecture': ['Arbiter', 'FF-Arbiter', 'Interpose', 'Lightweight', 'Permutation', 'XOR-Arbiter', 'XOR-FF-Arbiter'],
    'Temp_Min': [0.955, 0.980, 0.871, 0.928, 0.928, 0.926, 0.960],
    'Volt_Min': [0.861, 0.934, 0.689, 0.797, 0.797, 0.798, 0.880]
}

df = pd.DataFrame(data)
x = np.arange(len(df['Architecture']))
width = 0.35

fig, ax = plt.subplots(figsize=(10, 6))
# Plotting Min Reliability (Lower is worse)
ax.bar(x - width/2, df['Temp_Min'], width, label='Worst-case Temp Stress', color='#3498db', alpha=0.8)
ax.bar(x + width/2, df['Volt_Min'], width, label='Worst-case Voltage Stress', color='#e74c3c', alpha=0.8)

ax.set_ylabel('Minimum Reliability ($R_{min}$)', fontsize=14)
ax.set_title('Worst-Case Reliability Floor: Temperature vs. Voltage', fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(df['Architecture'], rotation=30, ha='right', fontsize=16)
ax.legend(loc='lower left')

# Focus on the failure zone
ax.set_ylim(0.6, 1.05) 
ax.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
